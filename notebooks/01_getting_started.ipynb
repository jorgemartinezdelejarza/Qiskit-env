# Getting Started with Qiskit

Welcome to The Road To Practitioner Program! This notebook introduces the basics of quantum computing with Qiskit.

## Step 1: Import Required Libraries

```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.primitives import Sampler
from qiskit.visualization import plot_histogram, circuit_drawer
import matplotlib.pyplot as plt
```

## Step 2: Create Your First Quantum Circuit

Let's create a simple circuit that creates a Bell state (also known as an entangled pair).

```python
# Create a quantum circuit with 2 qubits
qc = QuantumCircuit(2, 2, name='Bell State')

# Apply a Hadamard gate to qubit 0
qc.h(0)

# Apply a CNOT gate with qubit 0 as control and qubit 1 as target
qc.cx(0, 1)

# Measure both qubits
qc.measure([0, 1], [0, 1])

# Display the circuit
qc.draw(output='mpl')
```

## Step 3: Execute and Visualize Results

```python
# Create a sampler and run the circuit
sampler = Sampler()
result = sampler.run(qc, shots=1000).result()

# Get the counts from the result
quasi_dist = result.quasi_dists[0]
counts = {bitstring: int(count * 1000) for bitstring, count in quasi_dist.items()}

print("Measurement Results:")
print(f"{'Bitstring':<10} {'Count':<10} {'Probability':<15}")
print("-" * 35)
for bitstring, count in sorted(counts.items()):
    prob = count / 1000
    print(f"{bitstring:<10} {count:<10} {prob:<15.4f}")
```

## Step 4: Visualize the Results

```python
# Visualize the results
plot_histogram(counts)
```

## Understanding the Results

In a Bell state, we expect to see only two outcomes:
- `00`: Both qubits measure as 0
- `11`: Both qubits measure as 1

We should NOT see `01` or `10` because the qubits are entangled. This demonstrates quantum entanglement!

The equal probability of `00` and `11` shows the quantum superposition created by the Hadamard gate.
